```python
#Importar las librerías
import matplotlib.pyplot as plt
import networkx as nx
```

```python
#Creamos una instancia de la clase grafo
grafo = nx.Graph()
```

```python
grafo.add_edges_from(
    [
        ("SwitchA","PC1"),
        ("SwitchA","ImpresoraA"),
        ("SwitchA","RouterA"),
        ("RouterA","SwitchB"),
        ("RouterA","Servidor_1"),
        ("RouterA","SwitchB"),
        ("SwitchB","ImpresoraA"),
        ("SwitchB","PC2")
    ]
)
```

```python
#Mostrar el grafo
nx.draw(
    grafo,
    with_labels = True,
    node_size = 1000,
    node_color ="#ffc5d3"
)
plt.show
```
