```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
#Creamos el grafo
grafo=nx.DiGraph()
```

```python
grafo.add_weighted_edges_from([
    ("A", "B", 8),
    ("A", "D", 5),
    ("A", "E", 4),
    ("B", "F", 4),
    ("B", "C", 3),
    ("C", "F", 9),
    ("C", "G", 2),
    ("D", "E", 3),
    ("D", "H", 6),
    ("E", "I", 5),
    ("F", "K", 8),
    ("F", "E", 3),
    ("G", "L", 2),
    ("G", "K", 3),
    ("H", "I", 4),
    ("H", "M", 2),
    ("I", "J", 4),
    ("I", "M", 1),
    ("I", "N", 3),
    ("J", "O", 9),
    ("K", "J", 6),
    ("L", "K", 5),
    ("L", "P", 3),
    ("M", "N", 1),
    ("N", "O", 2),
    ("O", "P", 12),
    ("P", "K", 1)
])
```

```python
#Imprimimos el grafo
pos=nx.spring_layout(grafo, seed=12)
nx.draw(
    grafo,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
)
labels=nx.get_edge_attributes(grafo,"weight")
nx.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels=labels
)
plt.show()
```

```python
#Mostramos las rutas
rutas=list(
    nx.all_simple_paths(
        grafo,
        source="A",
        target="P"
    )
)
```

```python
#Imprimimos las rutas
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta{i}:{ruta}")
```

```python
### Calculamos el costo
def costo_ruta (grafo, ruta):
    costo=0
    for i in range (len(ruta)-1):
        costo += grafo [ruta[i]][ruta[i+1]]["weight"]
    return costo
```

```python
#Imprimimos el costo
for i, ruta in enumerate(rutas, start=1):
    costo=costo_ruta(grafo,ruta)
    print(f"Ruta{i}:{ruta}->{costo}unidades")
```

```python
#Elegir la ruta más corta
costos=[costo_ruta(grafo,ruta)
        for ruta in rutas
       ]
mejor_ruta=costos.index(min(costos))
```

```python
#Imprimir la mejor ruta
print(f"La mejor ruta es:{mejor_ruta+1}")
print(f"El camino corto es: {costos[mejor_ruta]}")
```

```python
#Algoritmo Dijkstra para obtener la ruta más corta
ruta_mas_corta=nx.dijkstra_path(grafo,source= "A", target= "P", weight= "weight")
tiempo_total=nx.dijkstra_path_length(grafo, source="A", target= "P", weight= "weight")
```

```python
#Imprimir
print(f"La ruta más corta es: {ruta_mas_corta}")
print(f"El tiempo total es: {tiempo_total} minutos")
```
