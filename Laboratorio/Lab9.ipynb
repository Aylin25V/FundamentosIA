```python
#Grafos dirigidos
import networkx as Snoopy
import matplotlib.pyplot as plt
```

```python
#Creamos el grafo
grafo=Snoopy.DiGraph()
```

```python
grafo.add_weighted_edges_from([
    ("A", "B", 5),
    ("A", "C", 2),
    ("B", "D", 4),
    ("C", "E", 5),
    ("C", "F", 6)
])
```

```python
#Imprimimos el grafo
pos=Snoopy.spring_layout(grafo, seed=12)
Snoopy.draw(
    grafo,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
)
labels=Snoopy.get_edge_attributes(grafo,"weight")
Snoopy.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels=labels
)
plt.show()
```

```python
#¿Puedo moverme de A a B?
print(grafo.has_edge("A", "B"))
```

```python
#¿Puedo moverme de B a A?
print(grafo.has_edge("B", "A"))
```

```python
#¿Puedo moverme de A a F?
print(grafo.has_edge("A", "F"))
```

```python
#Mostramos las rutas
rutas=list(
    Snoopy.all_simple_paths(
        grafo,
        source="A",
        target="F"
    )
)
```

```python
#Imprimimos las rutas
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta{i}:{ruta}")
```

```python
import networkx as Bingo
import matplotlib.pyplot as plt
```

```python
#Creamos el grafo
grafo=Bingo.DiGraph()
```

```python
grafo.add_weighted_edges_from([
    ("A", "B", 4),
    ("A", "C", 7),
    ("B", "D", 3),
    ("C", "D", 2),
    ("F", "G", 4),
    ("D", "G", 8),
    ("D", "E", 2),
    ("B", "E", 6),
    ("E", "G", 3),
    ("C", "F", 5)
])
```

```python
#Imprimimos el grafo
pos=Bingo.spring_layout(grafo, seed=12)
Bingo.draw(
    grafo,
    pos,
    with_labels=True,
    node_size=1500,
    arrows=True
)
labels=Bingo.get_edge_attributes(grafo,"weight")
Bingo.draw_networkx_edge_labels(
    grafo,
    pos,
    edge_labels=labels
)
plt.show()
```

```python
#Mostramos las rutas
rutas=list(
    Bingo.all_simple_paths(
        grafo,
        source="A",
        target="G"
    )
)
```

```python
#Imprimimos las rutas
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta{i}:{ruta}")
```

```python
### Calculamos el costo
def costo_ruta (grafo, ruta):
    costo=0
    for i in range (len(ruta)-1):
        costo += grafo [ruta[i]][ruta[i+1]]["weight"]
    return costo
```

```python
#Imprimimos el costo
for i, ruta in enumerate(rutas, start=1):
    costo=costo_ruta(grafo,ruta)
    print(f"Ruta{i}:{ruta}->{costo}unidades")
```

```python
#Elegir la ruta más corta
costos=[costo_ruta(grafo,ruta)
        for ruta in rutas
       ]
mejor_ruta=costos.index(min(costos))
```

```python
#Imprimir la mejor ruta
print(f"La mejor ruta es:{mejor_ruta+1}")
print(f"El camino corto es: {costos[mejor_ruta]}")
```
