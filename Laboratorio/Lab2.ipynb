# Problema: Realizar un sistema que identifique si una planta necesita riego para activar el sensor.

```python
print ("Sistema de riego de plantas")
porcentajeH = int(input("Ingresa el porcentaje de humedad de la tierra de tu plantita"))
print ("Porcentaje de humedad:", porcentajeH)
if porcentajeH < 40:
    print ("Riega tu plantita, irresponsable")
elif porcentajeH >= 40:
    print ("Tu plantita está hidratada, eres buen dueño")
    
```
