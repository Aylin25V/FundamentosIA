```python
#Importar las librerías
import matplotlib.pyplot as plt
import networkx as nx
```

```python
#Creamos una instancia de la clase graf
grafo = nx.Graph()
```

```python
grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("B","D"),
        ("D","F"),
        ("F","H"),
        ("F","E"),
        ("E","G"),
        ("C","G"),
        ("G","J"),
        ("F","H"),
        ("H","I"),
        ("I","J"),
    ]
)
```

```python
#Mostrar el grafo
nx.draw(
    grafo,
    with_labels = True,
    node_size = 1000,
    node_color ="#ffc5d3"
)
plt.show
```

```python
# Mostrar las rutas posibles del grafo
#Source: de dónde inicia
#Target: objetivo
rutas = list(
    nx.all_simple_paths(grafo, source="A", target="J")
)
```

```python
#Imprimir las rutas
print("Rutas encontradas:")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
```

```python
#Mostrar el total de pasos de cada ruta
pasos1 = len(rutas [0])-1
print("Pasos de la tura 1:", pasos1)
pasos2 = len(rutas [1])-1
print ("Pasos de la ruta 2:", pasos2)
pasos3 = len(rutas [2])-1
print ("Pasos de la ruta 3:", pasos3)
pasos4 = len(rutas [3])-1
print ("Pasos de la ruta 4:", pasos4)
```

```python
#Aplicamos la lógica proposicional
P = pasos1 < pasos2 #P: La mejor ruta es la 1
Q = pasos2 < pasos1 #Q: La mejor ruta es la 2
R = pasos3 < pasos4 #R: La mejor ruta es la 3
S = pasos4 < pasos3 #S: La mejor ruta es la 4
if P==True:
    print("P = True")
    print("La ruta 1 es mejor opción.")
elif Q==True:
    print("Q = True")
    print("La ruta 2 es mejor")
elif R==True:
    print("R = True")
    print("La ruta 3 es mejor")
elif S==True:
    print("S = True")
    print("La ruta 4 es mejor")
else:
    print("P = False")
    print("Q = False")
    print("Las dos rutas son igual de buenas, haz lo que quieras")
```
