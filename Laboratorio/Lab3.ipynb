### que debe decidir si puede iniciar un ataque contra el jugador.

#Proposiciones
P: Se encuentra dentro del rango
Q: El nivel de vida es suficiente
R: Existe municiones disponibles
S: Está visible
T: No existe amenaza cercana
U: No está aturdido

```python
# Definición de variables
rango = True #Sí está dentro del rango
nivelVida = True #Sí tiene vida suficiente
municiones = True #Sí cuenta con municiones disponibles
visible = True #Sí está visible el jugador
amenaza = False #No existe amenaza cercana
aturdido = False #No está aturdido
```

```python
puedeAtacar = (rango==True and visible==True and nivelVida==True and municiones==True and amenaza==True and aturdido==False)
```

```python
print("¿El agente enemigo puede atacar?", puedeAtacar)
```

```python
print (" == Agente Enemigo Inteligente ==")
print ("Development by Aylin")

rango = input ("El jugador se encuentra dentro del rango (s/n):").lower() == "s"
visible = input ("El jugador esta visible (s/n):").lower() == "s"
nivelVida = input ("El enemigo tiene suficiente vida(s/n):").lower() == "s"
municiones = input ("El enemigo tiene suficientes municiones(/n):").lower() == "s"
amenaza = input ("¿Existe una amenaza cercana? (s/n):").lower() == "n"
aturdido = input ("¿El enemigo no está aturdido? (s/n):").lower() == "n"

puedeAtacar = (rango==True and visible==True and nivelVida==True and municiones==True and amenaza==True and aturdido==True)

print ("Decisión del agente")
if puedeAtacar == True :
    print ("El agente puede atacar: It is your moment!!!")
else:
    print ("¡Escóndete!, no es momento de atacar")
```
